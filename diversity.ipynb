# Tree diversity distribution patterns

Notebook for reproducing Figure 1b in the paper GRASS as a Geocomputational Engine. 
It downloads and plots maps of tree diversity distribution patterns in the contiguous USA. 
See the tutorial [Estimating tree diversity distribution patterns using the r.series.diversity addon for GRASS GIS](https://ecodiv.earth/post/tree-species-diversity-distribution/) for the full workflow for calculating diversity indices from maps of tree species basal area.

**Requirements:** grass

**Dataset:** https://osf.io/uc6da

## Start GRASS

```python
# Import modules
import os
import sys
import subprocess
from pathlib import Path
import urllib.request
from zipfile import ZipFile
import numpy as np

# Find GRASS Python packages
sys.path.append(
  subprocess.check_output(
    ["grass", "--config", "python_path"],
    text=True
    ).strip()
  )

# Import GRASS packages
import grass.script as gs
import grass.jupyter as gj
from grass.tools import Tools

# Create and initialize GRASS project
gs.create_project("diversity", epsg="5070")
gj.init("diversity")
tools = Tools(overwrite=True)

# Print projection
projection = tools.g_proj(format="shell", flags="p").keyval
print(projection.get("name"))
```

## Download species richness

```python
# Download dataset
url = "https://osf.io/3v8ax/download"
filepath = "trees_usa_species_richness.zip"
try:
    urllib.request.urlretrieve(url, filepath)
except Exception as e:
    print(f"Error downloading file: {e}")

# Unarchive dataset
with ZipFile(filepath, 'r') as archive:
    archive.extractall()

# Delete archive
os.remove(filepath)
```

## Import species richness

```python
# Import
tools.r_import(input="trees_USA_species_richness.tif", output="species_richness")
```

## Visualize species richness

```python
# Visualize
m = gj.Map(width=800)
m.d_rast(map="species_richness")
m.d_legend(raster="species_richness", title="Species richness", at=(6, 8, 5, 30), labelnum=2, font="FiraSans-Regular", fontsize=18)
m.save("diversity-1.png")
m.show()
```
